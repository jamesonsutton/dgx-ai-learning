```python
import torch

x = torch.tensor([[-3.0],
                  [-2.0],
                  [-1.0],
                  [ 0.0],
                  [ 1.0],
                  [ 2.0],
                  [ 3.0]])
y = x ** 2

model = torch.nn.Sequential(
    torch.nn.Linear(1,8),
    torch.nn.ReLU(),
    # 8 times does: hiddenLayer1 = (h1×weight[0]) + (h2×weight[1]) + (h3×weight[2]) + (⋯) + (h8×weight[7]) + bias;
    torch.nn.Linear(8,8),
    torch.nn.ReLU(),
    torch.nn.Linear(8,1)
)

loss_fn = torch.nn.MSELoss()

optimizer = torch.optim.SGD(model.parameters(), lr = 0.0001)
```

```python
for epoch in range(100000):

    prediction = model(x)

    loss = loss_fn(prediction, y)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    if epoch % 1000 == 0:
        print("Epoch:", epoch, "Loss:", loss.item())
```

```python
test_x = torch.tensor([[-2.5],
                       [-1.5],
                       [-0.5],
                       [ 0.5],
                       [ 1.5],
                       [ 2.5]])

test_prediction = model(test_x)

print("Input:")
print(test_x)

print("\nAI prediction:")
print(test_prediction.detach())

print("\nActual x²:")
print(test_x ** 2)
```
